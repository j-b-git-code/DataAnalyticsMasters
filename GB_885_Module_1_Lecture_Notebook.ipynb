Imagine we have a factory with 90 employees, 10 of which are managers

```python
#assign variables
employees = 98
managers = 10
workers = employees - managers
```

```python
workers
```

```python
#we hire two more employees, but we have the same number of managers
employees = 100
workers
```

```python
#two workers retired, let's adjust our employee figure again
employees = 98
```

Lets say managers make \$150  per day. The rest make \$85.50.

```python
#assign wage variables
m_wage = 150
w_wage = 85.5
```

At this factory, one week is 5 working days. How much does the factory pay each week in wages?

```python
#calculate total wages paid per week

weekly_wage = 5* ((m_wage * managers) + (w_wage * workers))
weekly_wage
```

```python
#define daily operating expenses outside of labor, and daily revenue
daily_expenses = 1135.78
weekly_revenue = 60128.72
```

```python
#calculate weekly profit
#profit is the difference between the amount earned and the amount spent in running a business
weekly_profit = weekly_revenue - (daily_expenses*5 + weekly_wage)
weekly_profit
```

```python
#how much weekly profit are we making per employee?
weekly_profit / employees
```

The company wants to give each employee a raise, by dividing the remainder of the profit equally among them.

What will each employees raise be?

```python
#calculate remainder
weekly_profit % employees
```

```python
#how much will each employees' raise be?
(weekly_profit % employees) / employees
```

##Simple Functions

Let's imagine we work for a shipping company

```python
#write a print statement that welcoms our customers
print("Welcome to Starwing Shipping!")
```

```python
#create a variable for weight per kilogram
rate_per_kilo = 9.35
```

```python
#write an input function that asks our customer to enter the weight of their package and saves their answer as a variable.
cust_package_weight = input('How much does your package weigh in kilograms?')
cust_cost_estimate = cust_package_weight * rate_per_kilo
```

An error is returned: can't multiply sequence by non-int of type 'float'

All inputs are received as string data types. We can't multiply a string (customer's input) by a floating point number (shipping rate).

We will need to cast our input as a floating point number.

```python
#Let's try again
cust_package_weight = float(input('How much does your package weigh in kilograms?'))
cust_cost_estimate = cust_package_weight * rate_per_kilo
```

```python
#create a print statement that will give our customer a price quote
print(f'Your package will cost ${cust_cost_estimate} to ship.')
```

We don't want to express currency with more than two decimal points, so let's round our estimate.

```python
#let's try again
print(f'Your package will cost ${round(cust_cost_estimate, 2)} to ship.')
```
