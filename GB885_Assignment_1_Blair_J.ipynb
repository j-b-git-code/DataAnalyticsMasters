##Question 5
Calculate Mr. Cho’s Body Mass Index (BMI) using the following formula from the US Center for Disease Control:

BMI = 730 * Weight(lbs)/Height^2 (in)

What is Mr. Cho's BMI (round to the nearest hundredths place, or two place values to the right of the decimal)?

```python
#Instanciating and BMI
bmi = 0

#changing height from feet to inches by creating heightInInches and calculating
heightInInches = height * 12

#calculating BMI rounded to the hundredths place
bmi = round((703 * weight / (heightInInches ** 2)),2)

#output rounding to the reqired hundreths place
print(bmi)
```

##Question 6

Mr. Cho has been diagnosed with a bacterial infection. The formula for an adult amoxicillin dosage is: 10mg per Kilogram of patient weight, rounding to the nearest whole mg.

How many mg should be prescribed for a dosage of amoxicillin for Mr. Cho? Round to the nearest whole mg.

(Assume 1 pound = 0.45 kilograms.)


```python
#convert weight to weight in kg by multiplying weight by kg conversion
weightInKg = weight * .45

#calculating dose at 10 mg per kg to the whole kg
dose = round(weightInKg * 10)

print(dose)

```
